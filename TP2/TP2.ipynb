```python
import re
import doctest

md = """
# Cabeçalho1
## Cabeçalho2
### Cabeçalho3

**bold** **bold bold bold**
*italic* *italic italic italic*

1. Linha1
2. Linha2
3. Linha3

[texto texto](https://www.exemplo.pt)

![imagem imagem](https://www.imagem.xyz)
"""

def md_html(md):
    """
    Função que converte texto Markdown para texto HTML
    
    >>> print(md_html(md))
    <BLANKLINE>
    <h1>Cabeçalho1</h1>
    <h2>Cabeçalho2</h2>
    <h3>Cabeçalho3</h3>
    <BLANKLINE>
    <b>bold</b> <b>bold bold bold</b>
    <i>italic</i> <i>italic italic italic</i>
    <BLANKLINE>
    <ol>
    <li>Linha1</li>
    <li>Linha2</li>
    <li>Linha3</li>
    </ol>
    <a href="https://www.exemplo.pt">texto texto</a>
    <BLANKLINE>
    <img src="https://www.imagem.xyz" alt="imagem imagem"/>
    <BLANKLINE>
    """

    regexHeader = r"(#+) (.+)"
    regexBold = r"\*\*(.+?)\*\*"
    regexItalic = r"\*(.+?)\*"
    regexList = r"((?:\d\. .+\n?)+)"
    regexLink = r"\[(.+?)\]\((.+://[a-z0-9.-]+(/.*)?)\)"
    regexImg = r"!\[(.+?)\]\((.+://[a-z0-9.-]+(/.*)?)\)"

    def auxHeader(match):
        nivel = len(match.group(1))
        cabecalho = match.group(2)
        return f"<h{nivel}>{cabecalho}</h{nivel}>"

    def auxBold(match):
        texto = match.group(1)
        return f"<b>{texto}</b>"

    def auxItalic(match):
        texto = match.group(1)
        return f"<i>{texto}</i>"
    
    def auxList(match):
        texto = match.group(1)
        linhas = texto.splitlines()
        listaHTML = "<ol>\n"

        for linha in linhas:
            _, texto = linha.split(". ", 1)
            listaHTML += f"<li>{texto}</li>\n"

        listaHTML += "</ol>"
        return listaHTML

    def auxLink(match):
        texto = match.group(1)
        link = match.group(2)
        return f"<a href=\"{link}\">{texto}</a>"
        
    def auxImg(match):
        texto = match.group(1)
        img = match.group(2)
        return f"<img src=\"{img}\" alt=\"{texto}\"/>"

    md = re.sub(regexHeader, auxHeader, md)
    md = re.sub(regexBold, auxBold, md)
    md = re.sub(regexItalic, auxItalic, md)
    md = re.sub(regexList, auxList, md)
    md = re.sub(regexImg, auxImg, md)
    md = re.sub(regexLink, auxLink, md)

    return md

doctest.run_docstring_examples(md_html, globals())
```
