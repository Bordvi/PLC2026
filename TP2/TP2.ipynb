```python
import re
import doctest

md = """
# Cabeçalho1
## Cabeçalho2
### Cabeçalho3

**bold** **bold bold bold**
*italic* *italic italic italic*

1. Linha1
2. Linha2
3. Linha3

[texto texto texto](https://www.exemplo.pt)

![imagem imagem](https://www.imagem.xyz)
"""

def md_html(md):
    """
    Função que converte texto Markdown para texto HTML
    
    >>> print(md_html(md))
    <h1>Cabeçalho1</h1>
    <h2>Cabeçalho2</h2>
    <h3>Cabeçalho3</h3>

    <b>bold<b> <b>bold bold bold<b>
    <i>italic<i> <i>italic italic italic<i>

    <ol>
    <li>Linha1</li>
    <li>Linha2</li>
    <li>Linha3</li>
    </ol>
    
    Site: <a href="https://www.exemplo.pt">texto texto</a>

    Imagem: <img src="https://www.imagem.xyz" alt="imagem imagem"/>
    """

    regexHeader = r"(#+) (\w+)"
    regexBold = r"\*\*(.+?)\*\*"
    regexItalic = r"\*(.+?)\*"
    regexList = r"\d\. (.+)"
    regexLink = r"\[.+?\]\((.+://[a-z0-9.-]+(/.*)?)\)"
    regexImg = r"!\[.+?\]\((.+://[a-z0-9.-]+(/.*)?)\)"

    def auxHeader(match):
        nivel = len(match.group(1))
        cabecalho = match.group(2)
        return f"<h{nivel}>{cabecalho}</h{nivel}>"

    def auxBold(match):
        texto = match.group(1)
        return f"<b>{texto}</b>"

    def auxItalic(match):
        texto = match.group(1)
        return f"<i>{texto}</i>"
    
    def auxList(match):
        texto = match.group(1)
        return f"<ol>\n<li>{texto}</li>\n</ol>"

    return re.sub(regexList, auxList, md)

doctest.run_docstring_examples(md_html, globals())
```
